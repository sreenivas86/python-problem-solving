16. Prime number

```python
def prime(n:int):
    if n==1:
        return 'is Not Prime'
    for i in range(2, int(n**0.5)):
        if n%i==0:
            return 'is not prime'
    return 'is Prime'
```

```python
n=1
print(f'{n} {prime(n)}')
```

17 Reverse a Number

```python
def reverseNum(n:int):
    sign =-1 if n<0 else 1
    n=abs(n)
    rn=0
    while n>0:
        rem=n%10
        rn =(rn*10)+rem
        n= int(n/10)
    rn =rn*sign
    return rn
```

```python
n=-134
print(f'{n} is reversed {reverseNum(n)}')
```

```python
# second method
def reverseNum2(n:int):
    sign= -1 if n<0 else 1
    reverse= str(abs(n))[::-1]
    rn= sign*int(reverse)
    return rn
```

```python
n=-134
print(f'{n} is reversed {reverseNum2(n)}')
```

18. Palindrome

19.Count digits

```python
def countDigits(n:int):
    return len(str(abs(n)))
def countDigits2(n:int):
    count =0
    while n>0:
        count =count +1
        n= int(n/10)
    return count
```

```python
print(countDigits(12345))
countDigits2(12345)
```

19 ArmStrong


```python
def isArmstrong(n:int):
    temp=n
    if n<0: return False
    count= countDigits2(n)
    armTotal=0
    while n>0:
        rem = n%10
        armTotal = armTotal + (rem ** count)
        n =int(n/10)
    return temp==armTotal
```
