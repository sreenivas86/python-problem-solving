3.count vowels

```python
def countVowels(a:str) -> int:
    vowels=['a','e','i','o','u']
    count= 0
    a=a.lower()
    for i in a:
        if i in vowels:
            count +=1
    return count

countVowels('sreenivas')
```

4. Count consonants

```python
def countConsonants(a:str) -> int:
    conso='BCDFGHJKLMNPQRSTVWXYZbcdfghjklmnpqrstvwxyz'
    count=0
    for i in a:
        if i in conso:
            count =count+1
            
    return count
countConsonants('sreenivas')
```

7. To Check a string is anagrom or not
Anagram:
An anagram is a word or phrase made by mixing up the letters of another word or phrase

```python
# sorted() method takes string and return a sorted list

def sort_of_str(s:str)-> str:
    return ''.join(sorted(s))

    
def isAnagram(a:str, b: str) -> bool:
    if len(a) != len(a):
        return False
    
    return sort_of_str(a) == sort_of_str(b)
```

```python
isAnagram("earth" ,"heart")
```

47. To remove duplicate characters

```python
def remove_dup(word:str) -> str:
    unique=''
    for a in word:
        if a not in unique:
            unique= unique+a
    return unique
```

```python
remove_dup('sreenivas')
```

48. First non-repeating charecter

```python
def first_non_repeat_char(word : str)-> str:
   
    for i in word:
        if word.count(i)==1:
            return i
    return ''
```

```python
first_non_repeat_char('sreenivas')
```

```python
first_non_repeat_char('abcabc')
```

49. charecter frequency

```python
def char_frequency(word:str) -> dict[str,int]:
    freq: dict[str,int]=dict()
    for i in word:
        if freq.get(i) is None:
            freq.update({i:1})
        else:
            count:int=freq.get(i)
            
            freq.update({i:count+1})
            
    return freq
```

```python
char_frequency('sreenivas')
```
