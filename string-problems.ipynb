# String problems

1. reverse a string


```python
# method1 
def reverseString(string:str)-> str:
    
    
    return string[::-1]

reverseString('sreenivas')
```

```python
def reverseString2(string:str) -> str:
    rev:str= ''
    for i in range (len(string)-1,0,-1):
        rev=rev+string[i]
    return rev

reverseString2('sreenivas')
```

2. reverse a word in sentence



```python
# reverse words in sentence
def reverseWordsInString(a:str) -> str:
    words= a.split(' ')
    print(' '.join(words))
    rev: str= ''
    for i in range(len(words)):
        rev =rev +" "+ reverseString(words[i])
    return rev
reverseWordsInString("abc sab aba bac")
```

3.count vowels

```python
def countVowels(a:str) -> int:
    vowels=['a','e','i','o','u']
    count= 0
    a=a.lower()
    for i in a:
        if i in vowels:
            count +=1
    return count

countVowels('sreenivas')
```

4. Count consonants

```python
def countConsonants(a:str) -> int:
    conso='BCDFGHJKLMNPQRSTVWXYZbcdfghjklmnpqrstvwxyz'
    count=0
    for i in a:
        if i in conso:
            count =count+1
            
    return count
countConsonants('sreenivas')
```
