1. Character frequency

```python
# charecter frequency of  word
def char_freq(word:str)-> dict[str,int]:
    freq: dict[str,int]=dict()
    for char in word:
        if freq.get(char) is None:
            freq.update({char:1})
        else:
            count=freq.get(char)
            freq.update({char:count+1})
    return freq
```

```python
char_freq('sreenivas reddy')
```

```python
def word_freq(sentence: str)-> dict[str,int]:
    freq: dict[str,int]=dict()
    for word in sentence.split(' '):
        if freq.get(word) is None:
            freq.update({word:1})
        else:
            count =freq.get(word)
            freq.update({word: count+1})
    return freq
```

```python
word_freq('abc abc sree reddy ni sree')
```

54. Find Maximum Value
Goal

Find the largest value stored in a dictionary.

This becomes useful when you want:

Top 5 students, 
Most frequent words, 
Most common errors, 
Highest-selling products, 
Most-used API endpoints 

```python
def maximum_value(valueDict: dict[str,int]) -> dict[str,int]:
   
    max_name = next(iter(valueDict))
    max_num: int= valueDict.get(max_name)
    
    for name,number in valueDict.items():
        if number>max_num:
            max_num=number
            max_name=name
        
    
    return {max_name:max_num}
    
```

```python
maximum_value({
    "sreenivas": 80,
    "ravi": 95,
    "kiran": 75,
    'vishnu':99
})
```

```python
def minimum_value(value_dict:dict[str,int])-> dict[str,int]:
    min_name=next(iter(value_dict))
    min_value:int= value_dict.get(min_name)
    
    for name,value in value_dict.items():
        if min_value> value:
            min_name=name
            min_value=value
    
    return {min_name:min_value}
    
```

# 51. Character Frequency

### Goal

Count how many times each character occurs in a string.

Example:

`banana`

Think of the dictionary as:

| Key | Value |
| --- | ----: |
| `b` |     1 |
| `a` |     3 |
| `n` |     2 |

So the general pattern is:

**character → count**

### How to think

For every character:

1. Check whether it already exists.
2. If it doesn't exist, start its count at 1.
3. If it exists, increase its count.

### Pattern

**Frequency counting using a dictionary**

This is one of the most important patterns in Python.

# 52. Word Frequency

This is basically the same idea as character frequency, but now the unit is a **word** instead of a character.

### Example

Sentence:

`python is easy and python is powerful`

Frequency:

| Word     | Count |
| -------- | ----: |
| python   |     2 |
| is       |     2 |
| easy     |     1 |
| and      |     1 |
| powerful |     1 |

The dictionary becomes:

**word → number of occurrences**

### Difference from #51

| Problem             | Key       |
| ------------------- | --------- |
| Character frequency | Character |
| Word frequency      | Word      |

The underlying pattern is exactly the same.

### Real-world connection

This same pattern can be used for:

* Log level frequency
* HTTP status frequency
* Error message frequency
* File extension frequency
* API endpoint frequency

For example:

**status code → count**

`200 → 1500`
`404 → 73`
`500 → 12`

That's why this is useful for Data Engineering.

# 53. Merge Dictionaries

### Goal

Combine two dictionaries into one.

For example:

Dictionary A:

`name → Sreenivas`
`age → 25`

Dictionary B:

`city → Bengaluru`
`role → Engineer`

After merging:

`name → Sreenivas`
`age → 25`
`city → Bengaluru`
`role → Engineer`

### The important question

What happens if **both dictionaries contain the same key?**

Example:

Dictionary A:

`name → Sreenivas`

Dictionary B:

`name → Reddy`

You need a rule for the conflict.

Possible rules:

* Dictionary B wins
* Dictionary A wins
* Add the values
* Keep both values
* Raise an error

In basic Python problems, the common assumption is:

**The value from the later dictionary replaces the earlier value.**

### Pattern

**Dictionary combination + duplicate-key handling**

This is important because real data sources frequently contain overlapping keys.

**Dictionary traversal + comparison**

# 56. Sort Dictionary by Value

### Goal

Arrange dictionary entries according to their **values**, rather than their keys.

Example:

Original:

| Student | Marks |
| ------- | ----: |
| Ravi    |    78 |
| Suresh  |    91 |
| Arun    |    84 |

Sort by marks ascending:

| Student | Marks |
| ------- | ----: |
| Ravi    |    78 |
| Arun    |    84 |
| Suresh  |    91 |

Descending:

| Student | Marks |
| ------- | ----: |
| Suresh  |    91 |
| Arun    |    84 |
| Ravi    |    78 |

### Important concept

Normally, dictionaries are associated with:

**key → value**

But here you are saying:

> "I don't care about the key ordering. I want to order the entries based on the value."

So the sorting criterion becomes the **value**.

### Pattern

**Dictionary → sorting → value as sorting criterion**

This becomes useful when you want:

* Top 5 students
* Most frequent words
* Most common errors
* Highest-selling products
* Most-used API endpoints

# 57. Invert Dictionary

### Goal

Swap the keys and values.

Original:

| Key  | Value          |
| ---- | -------------- |
| `IN` | India          |
| `US` | United States  |
| `UK` | United Kingdom |

After inversion:

| Key            | Value |
| -------------- | ----- |
| India          | IN    |
| United States  | US    |
| United Kingdom | UK    |

So:

**key → value**

becomes:

**value → key**

### Very important problem

What if two keys have the same value?

Example:

`Ravi → 90`

`Suresh → 90`

If you invert it, you would get:

`90 → ?`

Which student should be stored?

You can't keep both in a normal dictionary because dictionary keys must be unique.

You therefore need to decide whether to:

* keep the last one
* keep the first one
* store a list of students

For example conceptually:

`90 → [Ravi, Suresh]`

### Pattern

**Swap key/value + handle duplicate values**

This is an excellent problem for understanding dictionary limitations.

# 58. Group Words by Length

This is a very important **grouping pattern**.

### Goal

Group words according to their length.

Example:

`cat dog apple banana sun`

Lengths:

* cat → 3
* dog → 3
* apple → 5
* banana → 6
* sun → 3

Result conceptually:

| Length | Words         |
| -----: | ------------- |
|      3 | cat, dog, sun |
|      5 | apple         |
|      6 | banana        |

The dictionary becomes:

**length → list of words**

### Important pattern

This is different from frequency.

Frequency:

**word → count**

Grouping:

**category → collection**

Here:

**word length → words**

### This pattern is extremely important

You will encounter:

**Group by department**

`department → employees`

**Group by country**

`country → users`

**Group by date**

`date → records`

**Group by status**

`status → transactions`

This is basically the Python equivalent of thinking about **SQL `GROUP BY`**.
