## Dictionary problems
1. Character frequency

```python
# charecter frequency of  word
def char_freq(word:str)-> dict[str,int]:
    freq: dict[str,int]=dict()
    for char in word:
        if freq.get(char) is None:
            freq.update({char:1})
        else:
            count=freq.get(char)
            freq.update({char:count+1})
    return freq
```

```python
char_freq('sreenivas reddy')
```

52. word freqency

Real-world connection

This same pattern can be used for:

Log level frequency,
HTTP status frequency,
Error message frequency,
File extension frequency,
API endpoint frequency,

```python
def word_freq(sentence: str)-> dict[str,int]:
    freq: dict[str,int]=dict()
    for word in sentence.split(' '):
        if freq.get(word) is None:
            freq.update({word:1})
        else:
            count =freq.get(word)
            freq.update({word: count+1})
    return freq
```

```python
word_freq('abc abc sree reddy ni sree')
```

54. Find Maximum Value
Goal

Find the largest value stored in a dictionary.

This becomes useful when you want:

Top 5 students, 
Most frequent words, 
Most common errors, 
Highest-selling products, 
Most-used API endpoints 

```python
def maximum_value(valueDict: dict[str,int]) -> dict[str,int]:
   
    max_name = next(iter(valueDict))
    max_num: int= valueDict.get(max_name)
    
    for name,number in valueDict.items():
        if number>max_num:
            max_num=number
            max_name=name
        
    
    return {max_name:max_num}
    
```

```python
maximum_value({
    "sreenivas": 80,
    "ravi": 95,
    "kiran": 75,
    'vishnu':99
})
```

55. find minimum value

```python
def minimum_value(value_dict:dict[str,int])-> dict[str,int]:
    min_name=next(iter(value_dict))
    min_value:int= value_dict.get(min_name)
    
    for name,value in value_dict.items():
        if min_value> value:
            min_name=name
            min_value=value
    
    return {min_name:min_value}
    
```

```python
minimum_value({
    "sreenivas": 80,
    "ravi":75,
    "kiran": 75,
    'vishnu':99,
    'chandu':75
})
```
