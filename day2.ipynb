### Bitwise Operators
- & -- Bitwise AND (Both bits true are true)
- | -- Bitwise OR (one bit is true is true )
- ^ -- XoR (Both bits same  return false)
- << Left shift
- '>> right shift

#### Left shift `<<`
    - Left shift operator moves all bits to left and fill the empty spaces on the right with zeros
    - example: 5<<1  bin(5)is 101 and add one zero at left 1010 dec(1010) is 10
    - formula is a<<b = a*(2**b)

#### right shift `>>`
    - Right shift operator move all bits to right and drops the bits fall of edge also add the zeros on left side
    - simple defination is removes bits from right
    - example: 5>>1 i.e bin(5) is 101 and remove 1 from right side 10 and dec(10) is 2
    - formula is a>>b = a/(2**b)
