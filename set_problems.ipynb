## set problems


Yes. These **61–70 problems are mainly about Python `set` operations**. I’ll explain the **problem, what you need to find, and the pattern**, without code.

# 61. Union

### Problem

You are given two sets and need to find **all elements that appear in either set**.

Example:

Set A:

`{1, 2, 3, 4}`

Set B:

`{3, 4, 5, 6}`

Union:

`{1, 2, 3, 4, 5, 6}`

### Think:

> "Give me everything from A and B, but don't repeat anything."

### Pattern

**A OR B**

Union is useful when you want to combine unique values from two datasets.

---

# 62. Intersection

### Problem

Find the elements that are **present in both sets**.

Example:

Set A:

`{1, 2, 3, 4}`

Set B:

`{3, 4, 5, 6}`

Intersection:

`{3, 4}`

Because `3` and `4` exist in both.

### Think:

> "What do these two sets have in common?"

### Pattern

**A AND B**

### Real-world example

Set A = students who attended Python training.

Set B = students who passed the test.

Intersection =

**Students who both attended and passed.**

---

# 63. Difference

### Problem

Find elements that are in the **first set but not in the second set**.

Example:

A:

`{1, 2, 3, 4}`

B:

`{3, 4, 5, 6}`

A − B:

`{1, 2}`

Because `1` and `2` exist in A but not B.

### Important

Difference is **directional**.

A − B is not necessarily the same as B − A.

B − A:

`{5, 6}`

### Think:

> "What does A have that B doesn't have?"

---

# 64. Symmetric Difference

This one is slightly different.

### Problem

Find elements that are present in **only one of the two sets**, but not both.

Example:

A:

`{1, 2, 3, 4}`

B:

`{3, 4, 5, 6}`

Symmetric difference:

`{1, 2, 5, 6}`

Why?

* `1, 2` → only in A → keep
* `3, 4` → in both → remove
* `5, 6` → only in B → keep

### Think:

> "Give me everything that is different between the two sets."

### Easy way to remember

**Union** → everything

**Intersection** → common

**Difference** → only from one side

**Symmetric difference** → only from either side, excluding common

---

# 65. Check Subset

### Problem

Determine whether **every element of one set exists in another set**.

Example:

A:

`{1, 2}`

B:

`{1, 2, 3, 4}`

A is a subset of B.

Because every element of A exists in B.

### Think:

> "Is A completely contained inside B?"

Another example:

A:

`{1, 5}`

B:

`{1, 2, 3, 4}`

A is **not** a subset of B because `5` doesn't exist in B.

### Important relationship

If:

**A ⊆ B**

then every element of A is also in B.

---

# 66. Check Superset

This is the reverse idea of subset.

### Problem

Determine whether one set contains **all elements of another set**.

Example:

A:

`{1, 2, 3, 4}`

B:

`{1, 2}`

A is a superset of B.

Because A contains everything that B contains.

### Think:

> "Does A completely contain B?"

### Relationship

If:

**A is a superset of B**

then:

**B is a subset of A**

So #65 and #66 are essentially the same relationship viewed from opposite directions.

---

# 67. Remove Duplicates

### Problem

You are given a collection containing repeated values and need to keep only unique values.

Example:

`[1, 2, 2, 3, 3, 3, 4]`

After removing duplicates:

`{1, 2, 3, 4}`

### Why is a set useful?

A set automatically stores **unique elements**.

So the main idea is:

> "Convert the collection into a set to eliminate duplicates."

### Important point

A set is primarily about **uniqueness**, not ordering.

If the problem requires maintaining the original order, you need to think beyond a simple set.

---

# 68. Count Unique Words

### Problem

Given a sentence, determine how many **different words** appear.

Example:

`python is easy and python is powerful`

Words:

`python, is, easy, and, python, is, powerful`

Unique words:

`python, is, easy, and, powerful`

Number of unique words:

**5**

### Think:

First:

**Extract words**

Then:

**Remove duplicates**

Then:

**Count the remaining words**

### Pattern

**Set + length**

This is a very common Data Engineering operation.

For example:

> How many unique users accessed the application?

You don't care how many times each user appeared.

You care about:

**distinct users**

---

# 69. Common Elements

This is closely related to **intersection**, but the problem is usually presented using lists or other collections rather than explicitly saying "set intersection."

### Problem

Find elements that occur in both collections.

Example:

List A:

`[1, 2, 3, 4, 5]`

List B:

`[3, 4, 5, 6, 7]`

Common elements:

`{3, 4, 5}`

### Think:

> "Which values appear in both?"

This is essentially an **intersection problem**.

### Why is this useful?

Suppose:

Dataset A contains:

`user1, user2, user3, user4`

Dataset B contains:

`user3, user4, user5`

Common users:

`user3, user4`

This is very common when comparing datasets.

---

# 70. Unique Elements

This problem can have **two meanings**, so read the wording carefully.

### Meaning 1: Remove duplicates

Example:

`[1, 2, 2, 3, 4, 4]`

Unique values:

`{1, 2, 3, 4}`

Here "unique" means:

> Values that appear at least once without duplicates.

---

### Meaning 2: Elements that appear exactly once

This is different.

Example:

`[1, 2, 2, 3, 4, 4]`

Elements appearing exactly once:

`{1, 3}`

Because:

* `1` → appears once
* `2` → appears twice
* `3` → appears once
* `4` → appears twice

### Very important distinction

**Unique values / distinct values:**

`{1, 2, 3, 4}`

**Values occurring exactly once:**

`{1, 3}`

When solving interview problems, make sure you understand which meaning the question intends.

# ⭐ The Set Mental Model

When you see a problem involving sets, ask:

| Question                           | Think                    |
| ---------------------------------- | ------------------------ |
| "Everything from A and B?"         | **Union**                |
| "What do they share?"              | **Intersection**         |
| "What does A have that B doesn't?" | **Difference**           |
| "What's in only one of them?"      | **Symmetric difference** |
| "Is A completely inside B?"        | **Subset**               |
| "Does A completely contain B?"     | **Superset**             |
| "Remove repeated values?"          | **Set / uniqueness**     |
| "How many different values?"       | **Set + count**          |
| "What appears in both?"            | **Intersection**         |
| "What appears exactly once?"       | **Frequency counting**   |

### One important connection to your previous problems

You have now moved from **dictionaries → sets**.

Remember:

**Dictionary**

> Used when you need **key → value**

Example:

`"error" → 25`

**Set**

> Used when you need **unique values**

Example:

`{"error", "warning", "info"}`

So the key question is:

> **Do I need to store information about each value, or do I only need to know which values exist uniquely?**

If you need **counts, mappings, or associated information → dictionary**.

If you need **uniqueness, membership, or set relationships → set**.
