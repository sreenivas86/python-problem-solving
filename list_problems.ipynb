3. Second largest number in a list


```python
# first method

def secondLargestNum(num:list[int]) ->int:
    if len(num)==1:
        return num[0]
    num.sort(reverse=True)
    
    return num[1]

# second method
def secondLargestNum1(num:list[int]) -> int:
    if len(num)==1:
        return num[0]
    if num[0]> num[1]:
        first=num[0]
        second= num[1]
    else:
        first=num[1]
        second= num[0]
    
    for i in range(1,len(num)):
        if first< num[i]:
            second= first
            first=num[i]
        if second< num[i] and first> num[i]:
            second=num[i]
            
    return second
```

```python
# test
n=[5,7,3,4,2,5,8,0,9,10]
secondLargestNum(n)
```

4. Second smallest Number

```python
# first method
def secondSmallestNum(num:list[int]) -> int:
    if len(num)==1:
        return num[0]
    num.sort()
    return num[1]

# second method

def secondSmallestNum1(num:list[int]) -> int:
    if len(num)==1:
        return num[0]
    if num[0]< num[1]:
        first=num[0]
        second= num[1]
    else:
        first=num[1]
        second= num[0]
    for i in range(2, len(num)):
        if first>num[i]:
            second=first
            first= num[i]
        if second> num[i] and first<num[i]:
            second=num[i]
    return second
```

```python
# test
n=[5,7,3,4,2,5,8,0,9,10]
secondSmallestNum(n)
```
