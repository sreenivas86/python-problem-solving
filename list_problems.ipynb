3. Second largest number in a list


```python
# first method

def secondLargestNum(num:list[int]) ->int:
    if len(num)==1:
        return num[0]
    num.sort(reverse=True)
    
    return num[1]

# second method
def secondLargestNum1(num:list[int]) -> int:
    if len(num)==1:
        return num[0]
    if num[0]> num[1]:
        first=num[0]
        second= num[1]
    else:
        first=num[1]
        second= num[0]
    
    for i in range(1,len(num)):
        if first< num[i]:
            second= first
            first=num[i]
        if second< num[i] and first> num[i]:
            second=num[i]
            
    return second
```

4. Second smallest Number

```python
# first method
def secondSmallestNum(num:list[int]) -> int:
    if len(num)==1:
        return num[0]
    num.sort()
    return num[1]

# second method

def secondSmallestNum1(num:list[int]) -> int:
    if len(num)==1:
        return num[0]
    if num[0]< num[1]:
        first=num[0]
        second= num[1]
    else:
        first=num[1]
        second= num[0]
    for i in range(2, len(num)):
        if first>num[i]:
            second=first
            first= num[i]
        if second> num[i] and first<num[i]:
            second=num[i]
    return second
```

```python
# test
n=[5,7,3,4,2,5,8,0,9,10]
secondSmallestNum(n)
```

5. Find duplicate values

```python
# find duplicate values
def duplicateValues(num:list[int]) :
    duplicate :set[int]=set()
    for i in num:
        if num.count(i)>1:
            duplicate.add(i) 
    return duplicate 


#2nd method
def duplicateValues2(num:list[int]) -> set[int]:
    seen :set[int]=set()
    duplicate : set[int]=set()
    for i in num:
        if i in seen:
            duplicate.add(i)
        else:
            seen.add(i)
    return duplicate
```

```python
nums=[4,5,33,3,6,5,3,4,6,1,1,3,3,3,4,2]
duplicateValues(num=nums)
```

6. remove duplicates

```python
# method 1 O(n)
def removeduplicates(nums:list[int]) -> list[int]:
    return list(set(nums))

# method 2 O(n^2)
def removeduplicates1(nums:list[int]) -> list[int]:
    uniques: list[int]=list()
    for num in nums:
        if num not in uniques:
            uniques.append(num)
    return uniques


# method 3 O(n) using set
def removeduplicates2(nums: list[int]) -> list[int]:
    unique:list[int]= list()
    seen: set[int]=set()
    for num in nums:
        if num not in seen:
            seen.add(num)
            unique.append(num)
            
    return unique
```

```python
nums = [1, 2, 3, 2, 4, 5, 3, 6]
removeduplicates(nums)
```

```python
nums = [1, 2, 3, 2, 4, 5, 3, 6]
removeduplicates1(nums=nums)
```

```python
nums = [1, 2, 3, 2, 4, 5, 3, 6]
removeduplicates2(nums)
```

7. left Rotate a list
A Left rotation moves elements one position to the left. the elements moves to the right

ex:
[1,2,3,4,5]
1 left rotation
[2,3,4,5,1]

```python
# slice method
def leftRotate(nums:list[int], k:int) -> list[int]:
    p=k%len(nums)
    return nums[p:]+nums[:p]

nums= [1,2,3,4,5]
leftRotate(nums=nums, k=2)
```

```python
# loop method

def leftRotate1(nums:list[int], k:int) ->list[int]:
    n=len(nums)
    k=k%n
    result: list[int]= [0]*n
    for i in range (n):
        result[(i-k)%n]=nums[i]
    return result

nums= [1,2,3,4,5]
leftRotate1(nums=nums, k=2)
```

8. right rotate a list
A right rotation moves element one position to the right. the elements moves to the front or left

ex:
[1,2,3,4,5]
1 right rotation
[5,1,2,3,4]

```python
# slice method
def rightRotate(nums: list[int], k:int) -> list[int]:
    k=k%len(nums)
    return nums[-k:]+ nums[:-k]

nums= [1,2,3,4,5]
rightRotate(nums=nums, k=2)
```

```python
# for loop

def rightRotate1(nums: list[int], k:int) -> list[int]:
    n=len(nums)
    k=k%n
    result =[0]*n
    for i in range(n):
        result[k-i]=nums[i]
    return result
nums= [1,2,3,4,5]
rightRotate(nums=nums, k=2)
```

9. To merge two sorted arrays
    ex: [2,4,6] + [1,3,7,9]= [1,2,3,4,7,9]

```python
# using sort() method
def merge_two_sorted_list(list1:list[int], list2:list[int]) -> list[int]:
    merge=list1+list2
    merge.sort()
    return  merge

a=[1,4,6,9]
b=[2,3,7,8,9,10]
merge_two_sorted_list(a,b)
```

```python
# custom method
def merge_two_sorted_list1(list1:list[int], list2:list[int]) -> list[int]:
    merge:list[int]=[]
    l1=len(list1)
    l2=len(list2)
    i=j=0
    while i<l1 and j<l2:
        if list1[i]< list2[j]:
            merge.append(list1[i])
            i=i+1
        else :
            merge.append(list2[j])
            j= j+1
    
    if i<l1:
        merge= merge+ list1[i:] 
    if j<l2:
        merge =merge + list2[j:]  
    return  merge
    
a=[1,4,6,9]
b=[2,3,7,8,9,10]
merge_two_sorted_list1(a,b)
```

10. Find the missing number.
[1,2,3,4,6] m=5

```python
# sum method
def find_missing_num(nums: list[int])-> int:
    size= len(nums)+1
    actual_total=(size*(size+1))/2
    total=0
    for i in nums:
        total =total +i
    return int(actual_total-total)
a=[1,2,3,4,6]
find_missing_num(a)
```
