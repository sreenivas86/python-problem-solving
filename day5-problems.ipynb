Problmes on Lists

1. find the largest element

```python
# first method
def largestElement(num:list[int]) -> int:
    num.sort(reverse=True)
    return num[0]

# second method
def largestElement1(num:list[int]) -> int:
    largest=num[0]
    for i in num:
        if i>largest:
            largest=i
            
    return largest
```

```python
num = [4,5,2,6,7,2,3,8,3,1]
print(largestElement(num))
largestElement1(num)
```

2 Find the smallest number


```python
# first method
def smallestNumber(num:list[int])->int:
    num.sort()
    return num[0]

# second method
def smallestNumber2(num:list[int])-> int:
    smallest=num[0]
    for i in num:
        if i<smallest:
            smallest =i
    return smallest
```

```python
num = [4,5,2,6,7,2,3,8,3,1]
print(smallestNumber(num))
smallestNumber2(num)
```
