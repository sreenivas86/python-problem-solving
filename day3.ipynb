## Literals
- In python, A Literal is a fixed and raw value written into the source code that requires no  runtime computation
    - Boolean
    - Numeric
    - String
    - Special
    - Collection


###### String literals
- Single or double quates are used to define a string
- There is two types of string literals in python
    - single line string:
        - ex: text='hello'
    - multiline string:
        - adding a back slash at the end of a line
        - ex: text= " hello \
                    sreenivas "

- String types:
    - Raw string: Prefixed with an `r` or `R` to treat backslash as a literal
    - Formatted String: prefixed with `f`or `F` to evaluate embedded python expressions at runtime.
    - Bytes literals: Prefixed with `b` or `B` to generate immutable sequence of bytes instead of text charecters

```python
# exaple of those literals
text1= 'sreenivas'
text2 = "sreenivas \
    reddy\
        sreenu\
        "
print( text1)
text2
```
